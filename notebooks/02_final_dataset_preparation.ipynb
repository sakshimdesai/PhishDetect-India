```python
url_path = (
    "../data/raw/"
    "PhiUSIIL_Phishing_URL_Dataset.csv"
)

url_data = pd.read_csv(url_path)

print("Shape:", url_data.shape)

print("\nColumns:")
print(url_data.columns.tolist())

display(url_data.head())
```

```python
print("Label distribution:")
print(url_data["label"].value_counts())

print("\nLabel percentages:")
print(url_data["label"].value_counts(normalize=True) * 100)

print("\nMissing values:")
print(url_data.isnull().sum().sort_values(ascending=False).head(15))

print("\nDuplicate URLs:")
print(url_data["URL"].duplicated().sum())

print("\nDuplicate rows:")
print(url_data.duplicated().sum())
```

```python
print("Examples for label 0:")
display(
    url_data[url_data["label"] == 0][
        ["URL", "Domain", "IsDomainIP", "IsHTTPS", "URLLength", "label"]
    ].head(10)
)

print("\nExamples for label 1:")
display(
    url_data[url_data["label"] == 1][
        ["URL", "Domain", "IsDomainIP", "IsHTTPS", "URLLength", "label"]
    ].head(10)
)
```

```python
duplicate_urls = (
    url_data[url_data["URL"].duplicated(keep=False)]
    .sort_values("URL")
)

print("Rows involved in duplicate URLs:", len(duplicate_urls))

print("\nDuplicate URLs with multiple labels:")
label_conflicts = (
    duplicate_urls.groupby("URL")["label"]
    .nunique()
)

print(label_conflicts.value_counts())

print("\nNumber of conflicting URLs:")
print((label_conflicts > 1).sum())
```
